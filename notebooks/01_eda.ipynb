# Group EDA

Set up your private group repository first. Place the authorized `train.csv` and `test.csv` in local `data/` (ignored by Git). This notebook has no precomputed competition output.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from competition_eda import load_train, validate_test, full_calendar_days, create_plots

CELL = "Cell_177"  # Change this to your chosen cell.
train = load_train(ROOT / "data" / "train.csv")
validate_test(ROOT / "data" / "test.csv")
train[["timestamp", CELL]].head()
```

## One complete, reproducible plot

```python
REPORTS = ROOT / "reports"
REPORTS.mkdir(exist_ok=True)
fig = plt.figure(figsize=(10, 4), layout="constrained")
ax = fig.gca()
ax.plot(train["timestamp"], train[CELL], linewidth=1)
ax.set(title=f"{CELL}: training history", xlabel="Recorded timestamp",
       ylabel="Anonymized uplink count per 5-minute window")
ax.grid(True, alpha=0.2)
fig.savefig(REPORTS / "first_plot.png", dpi=220, bbox_inches="tight")
display(fig)
plt.close(fig)
```
